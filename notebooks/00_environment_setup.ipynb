| 종류 | 용도 | 전환 단축키 |

### 필수 단축키

| 단축키 | 동작 |
|--------|------|
| `Shift + Enter` | 셀 실행 후 다음 셀 이동 |

| `Ctrl + Z` | 실행 취소 |
| `00` (숫자 영) | 커널 재시작 |

### 커널(Kernel) 관리

- **커널**: Python 코드를 실제로 실행하는 프로세스
- 메뉴 → **Kernel → Restart & Run All**: 처음부터 전체 실행
- 메뉴 → **Kernel → Restart & Clear Output**: 출력 초기화 후 재시작
- `In [*]` 표시는 실행 중, `In [숫자]`는 완료

## 4️⃣ 데이터셋 다운로드

### KAMP (한국스마트제조AI) 공개 데이터

이 교육과정은 **KAMP(https://www.kamp-ai.kr)** 의 공개 제조 데이터를 활용합니다.

| 데이터셋 | 설명 | 사용 노트북 |
|---------|------|-------------|
| CNC 공정 데이터 | 가공 조건별 불량 여부 | 01, 02, 03 |
| 설비 진동 데이터 | 베어링 상태 분류 | Track A-1 |
| 용접 품질 이미지 | 정상/불량 이진 분류 | Track B-1 |

### 샘플 데이터 자동 다운로드

````python
import os
import urllib.request
import numpy as np
import pandas as pd

# 데이터 저장 디렉토리 생성
DATA_DIR = '../data'
os.makedirs(DATA_DIR, exist_ok=True)
print(f'📁 데이터 디렉토리: {os.path.abspath(DATA_DIR)}')

# ── 샘플 데이터 생성 (KAMP 접속 불가 시 사용) ──────────────────────
def create_sample_cnc_data(n=500):
    """CNC 공정 샘플 데이터 생성"""
    np.random.seed(42)
    df = pd.DataFrame({
        'spindle_speed':   np.random.randint(800, 2000, n),
        'feed_rate':       np.random.randint(100, 500, n),
        'depth_of_cut':    np.round(np.random.uniform(0.5, 3.0, n), 2),
        'tool_wear':       np.round(np.random.uniform(0, 1, n), 3),
        'vibration_rms':   np.round(np.abs(np.random.randn(n)) * 0.3 + 0.5, 3),
        'temperature':     np.round(np.random.normal(75, 10, n), 1),
        'coolant_flow':    np.round(np.random.uniform(2, 8, n), 1),
    })
    # 불량 레이블: 공구마모 높거나 진동 크면 불량 확률 높음
    defect_prob = (df['tool_wear'] > 0.7).astype(float) * 0.6 + \
                  (df['vibration_rms'] > 0.9).astype(float) * 0.4
    df['defect'] = (np.random.rand(n) < defect_prob).astype(int)
    return df

def create_sample_vibration_data(n=1000):
    """설비 진동 샘플 데이터 생성"""
    np.random.seed(42)
    # 정상(0), 내륜결함(1), 외륜결함(2), 볼결함(3)
    labels = np.repeat([0, 1, 2, 3], n // 4)
    noise_level = [0.1, 0.3, 0.5, 0.4]
    rows = []
    for lbl in labels:
        base = np.random.randn(20) * noise_level[lbl]
        if lbl > 0:  # 결함 주파수 성분 추가
            base[lbl * 3] += lbl * 0.8
        rows.append(list(base) + [lbl])
    cols = [f'feat_{i}' for i in range(20)] + ['fault_type']
    return pd.DataFrame(rows, columns=cols)

# 샘플 데이터 저장
cnc_path = os.path.join(DATA_DIR, 'sample_kamp_cnc.csv')
vib_path = os.path.join(DATA_DIR, 'sample_kamp_vibration.csv')

if not os.path.exists(cnc_path):
    df_cnc = create_sample_cnc_data()
    df_cnc.to_csv(cnc_path, index=False)
    print(f'✅ CNC 샘플 데이터 생성: {cnc_path} ({len(df_cnc)}행)')
else:
    print(f'✅ CNC 데이터 기존 파일 사용: {cnc_path}')

if not os.path.exists(vib_path):
    df_vib = create_sample_vibration_data()
    df_vib.to_csv(vib_path, index=False)
    print(f'✅ 진동 샘플 데이터 생성: {vib_path} ({len(df_vib)}행)')
else:
    print(f'✅ 진동 데이터 기존 파일 사용: {vib_path}')

print()
print('📥 KAMP 실제 데이터가 필요한 경우:')
print('   https://www.kamp-ai.kr → 데이터셋 → 스마트제조 검색')
print('   다운로드 후 ../data/ 폴더에 저장하세요')
````

````python
# 데이터 확인
df_cnc = pd.read_csv(cnc_path)
print('=== CNC 공정 데이터 미리보기 ===')
print(f'크기: {df_cnc.shape[0]}행 × {df_cnc.shape[1]}열')
print(f'불량률: {df_cnc["defect"].mean():.1%}')
display(df_cnc.head(3))
print()
print('=== 기술 통계 ===')
display(df_cnc.describe().round(2))
````

## 5️⃣ 환경 최종 점검

아래 셀을 실행해서 **모든 항목이 ✅** 인지 확인하세요.  
❌ 가 있으면 해당 라이브러리를 `pip install 라이브러리명` 으로 설치하세요.

````python
import importlib

checks = [
    # (import명, 표시명, 최소버전)
    ('numpy',       'NumPy',       '1.23'),
    ('pandas',      'Pandas',      '1.5'),
    ('matplotlib',  'Matplotlib',  '3.6'),
    ('seaborn',     'Seaborn',     '0.12'),
    ('scipy',       'SciPy',       '1.9'),
    ('sklearn',     'Scikit-learn','1.2'),
    ('joblib',      'Joblib',      '1.2'),
    ('tqdm',        'tqdm',        '4.0'),
    ('torch',       'PyTorch',     '2.0'),
    ('torchvision', 'TorchVision', '0.15'),
    ('shap',        'SHAP',        '0.41'),
]

print('=' * 55)
print('🔍 환경 점검 결과')
print('=' * 55)

all_ok = True
for mod, name, min_ver in checks:
    try:
        m = importlib.import_module(mod)
        ver = getattr(m, '__version__', '?')
        # PyCaret은 별도 처리
        print(f'  ✅ {name:<15} {ver}')
    except ImportError:
        print(f'  ❌ {name:<15} 미설치  →  pip install {mod}')
        all_ok = False

# PyCaret 별도 확인 (설치 시간 오래 걸림)
try:
    import pycaret
    print(f'  ✅ {"PyCaret":<15} {pycaret.__version__}')
except ImportError:
    print(f'  ⚠️  {"PyCaret":<15} 미설치  →  pip install pycaret  (선택사항, 느림)')

print('=' * 55)
if all_ok:
    print('🎉 모든 필수 라이브러리 설치 완료! 실습을 시작하세요.')
else:
    print('⚠️  위 ❌ 항목을 설치 후 이 셀을 다시 실행하세요.')
````
